# Exploratory Data Analysis (EDA)

This notebook performs the first Day 1 dataset checks for the Rajora Career Engine project.

We will:

1. Load the Excel datasets from `data/raw/`.
2. Check each dataset's shape.
3. Display column names.
4. Count missing values.
5. Review class distribution for target-like columns such as `career_goal` and `recommended_career_1`.

## 1. Import libraries

Pandas is used to load and inspect tabular data. Pathlib helps us work with file paths in a clean way.

```python
from pathlib import Path

import pandas as pd

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

## 2. Locate datasets

The datasets are stored in `data/raw/`. We keep raw datasets outside Git tracking, but they can be used locally for analysis.

```python
DATA_DIR = Path("../data/raw")
excel_files = sorted(DATA_DIR.glob("*.xlsx"))

print(f"Found {len(excel_files)} Excel dataset(s):")
for file_path in excel_files:
    print(f"- {file_path.name}")
```

## 3. Load datasets

Each workbook is loaded from its first sheet. We store the loaded DataFrames in a dictionary so we can inspect them one by one.

```python
datasets = {}

for file_path in excel_files:
    excel_file = pd.ExcelFile(file_path)
    first_sheet = excel_file.sheet_names[0]
    df = pd.read_excel(file_path, sheet_name=first_sheet)
    datasets[file_path.name] = df
    print(f"Loaded {file_path.name} | sheet: {first_sheet} | shape: {df.shape}")
```

## 4. Shape and columns

Shape tells us the number of rows and columns. Columns tell us what features are available for model training.

```python
for name, df in datasets.items():
    print("=" * 100)
    print(name)
    print(f"Shape: {df.shape[0]} rows x {df.shape[1]} columns")
    print("Columns:")
    for column in df.columns:
        print(f"- {column}")
```

## 5. Missing values

Missing values show where data cleaning is needed before model training.

```python
for name, df in datasets.items():
    missing_values = df.isna().sum().sort_values(ascending=False)
    missing_values = missing_values[missing_values > 0]

    print("=" * 100)
    print(name)
    print(f"Total missing values: {int(df.isna().sum().sum())}")

    if missing_values.empty:
        print("No missing values found.")
    else:
        display(missing_values.to_frame("missing_count"))
```

## 6. Class distribution

Class distribution helps us understand whether the target labels are balanced or imbalanced. Here we inspect `career_goal` and `recommended_career_1` when they exist.

```python
target_columns = ["career_goal", "recommended_career_1"]

for name, df in datasets.items():
    print("=" * 100)
    print(name)

    for target_column in target_columns:
        if target_column in df.columns:
            print(f"\nClass distribution for {target_column}:")
            display(df[target_column].value_counts(dropna=False).to_frame("count"))
        else:
            print(f"\nColumn not found: {target_column}")
```

## 7. Quick preview

Previewing rows helps confirm that the dataset loaded correctly and the columns look meaningful.

```python
for name, df in datasets.items():
    print("=" * 100)
    print(name)
    display(df.head())
```
